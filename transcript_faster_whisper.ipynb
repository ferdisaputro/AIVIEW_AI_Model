```python
from faster_whisper import WhisperModel

model_size = "large-v3"

# Run on GPU with FP16
model_fp16 = WhisperModel('turbo', device="cuda", compute_type="float16")

audio_path = "datasets/tips_interview_1080p.mp4"
segments_fp16, info_fp16 = model_fp16.transcribe(audio_path, beam_size=5)

print("Detected language '%s' with probability %f" % (info_fp16.language, info_fp16.language_probability))
 
# for segment in segments_fp16:
# 	print("[%.2fs -> %.2fs] %s" % (segment.start, segment.end, segment.text))
```

```python
# or run on GPU with INT8
model_int8 = WhisperModel(model_size, device="cuda", compute_type="int8_float16")

segments_int8, info_int8 = model_int8.transcribe(audio_path, beam_size=5)

print("Detected language '%s' with probability %f" % (info_int8.language, info_int8.language_probability))

for segment in segments_int8:
	print("[%.2fs -> %.2fs] %s" % (segment.start, segment.end, segment.text))
```
